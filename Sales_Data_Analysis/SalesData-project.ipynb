### Section 2 Data Cleaning
   - Handle the null in Salesperson column — think carefully, should you fill or drop and why? Write your reason in a comment
   - Handle the nulls in Units_Sold — which method is correct here and why?
   - Create a new column called Total_Revenue by calculating: Units_Sold × Unit_Price
   - Now apply the discount to get Final_Revenue: Total_Revenue minus the discount percentage

```python
df.Salesperson.mode()
```

```python
#Filling missing values in Salesperson column with mode becuase the dataset is small and it's categorical column so it's the right choice.

df.Salesperson = df.Salesperson.fillna(df.Salesperson.mode()[0])
```

```python
#Checking if there is any outlier in this column before filling missing values.
import seaborn as sns

sns.boxplot(df.Units_Sold)
```

```python
#Checking mean of Units_Sold column.
df.Units_Sold.mean()
```

```python
#Filling the missing values with mean in Units_Sold column becuase there is no outlier so the distribution will be fine.

df.Units_Sold = df.Units_Sold.fillna(df.Units_Sold.mean()) 
```

```python
#Creating new column Total_Revenue by calculating units_sold * units_price
df['Total_Revenue'] = df.Units_Sold * df.Unit_Price
```

```python
df.head()
```

```python
#Updating Discount_% column with float values.
df['Discount_%'] = df['Discount_%']/100
df.head()
```

```python
#Calculating Discount_amount by multiplying Total_Revenue with Discount_%
discount_amt = df['Total_Revenue'] * df['Discount_%'] 
discount_amt
```

```python
#Creating new column Final_Revenue by subtracting Discount_amount from Total_Revenue
df['Final_Revenue'] = df.Total_Revenue - discount_amt 
df.head()
```

```python
df['Returned'] = df['Returned'].replace('No', 0).replace('Yes', 1).astype(int)
```
