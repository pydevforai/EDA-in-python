```python
def decide_tax_percentage(price):
    if price >= 20:
        return 15
    elif price >= 15:
        return 10
    elif price >= 10:
        return 5
    else:
        return 3    
```

```python
df2['Tax%'] = df2.Price.apply(decide_tax_percentage)
df2
```

```python
#Inline function when we need it just one time 
df2['Tax%'] = df2.Price.apply(lambda x: 15 if x >= 20 else 10 if x >= 15 else 5 if x >= 10 else 3)
df2
```

```python
df2['Tax_amt'] = df2.Price.apply(lambda x: x*.15 if x >= 20 else x*.10 if x >= 15 else x*.05 if x >= 10 else x*.03)
df2
```

```python
df2['Net'] = df2['Price'] + df2['Tax_amt']
df2
```

### Detecting and Filtering Outliers

```python
import numpy as np

def outlier_detector(col):
    q1 = np.percentile(col, 25)
    q3 = np.percentile(col, 75)
    IQR = q3-q1
    lower_boundary = q1-1.5*IQR
    upper_boundary = q3+1.5*IQR
    return lower_boundary, upper_boundary
```

```python
lb, ub = outlier_detector(df7.Salary)
lb, ub 
```

```python
outlier_detector(df7.Age)
```
